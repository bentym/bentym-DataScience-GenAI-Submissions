## Task 2 – Separate Predictors and Target

Set `loan_status` as the target (Y) and use the remaining columns as predictors (X). Check the dimensions of both.

```python
X = df.drop(columns=['loan_status'])
y = df['loan_status']

print('X shape:', X.shape)
print('y shape:', y.shape)
```

## Task 3 – Split the Data

Create training and test sets, using 80% of the observations for training and 20% for testing. Preserve the class proportions and use a fixed random seed so your results are reproducible. Check the resulting shapes.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

print('X_train shape:', X_train.shape)
print('X_test shape:', X_test.shape)
print('y_train shape:', y_train.shape)
print('y_test shape:', y_test.shape)
```

## Task 4 – Scale the Predictors

Scale the predictors so they are on a comparable scale. Fit the scaling procedure using **training data only**, then apply the same transformation to the test data. Why should the test set not be used to fit the scaler?

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Fit on training data and transform
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the fitted scaler
X_test_scaled = scaler.transform(X_test)

# Verify the shapes
print('X_train_scaled shape:', X_train_scaled.shape)
print('X_test_scaled shape:', X_test_scaled.shape)
```

### Why the test set should not be used to fit the scaler:
Fitting the scaler on the test set (or the entire dataset before splitting) causes **data leakage** (or information leakage).

When we fit a scaler, it calculates statistics such as the mean and standard deviation of the features. If we use the test set to calculate these statistics, information from the test set leaks into the training process. The test set is meant to act as an unseen, independent dataset to evaluate how well our model generalizes to completely new data. By fitting the scaler on training data only, we ensure that our preprocessing and modeling steps mimic a real-world scenario where incoming new data is scaled purely using the historical distributions learned from the training set.
